## Feature Engineering

```python
# 1- Date of Journey
df['Day'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df['Day'] = df['Day'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
```

```python
df.info()
```

```python
# drop date of journey
df.drop('Date_of_Journey', axis=1, inplace= True)
```

```python
# 2- Arrival Time
df['Arrival_Time']= df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python

df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_minute'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Arrival_hour'] = df['Arrival_hour'].astype(int)
df['Arrival_minute'] = df['Arrival_minute'].astype(int)
```

```python
df.drop('Arrival_Time', axis=1, inplace= True)
```

```python
df.head(2)
```

```python
df['Dep_hour'] = df['Dep_Time'].str.split(':').str[0]
df['Dep_minute'] = df['Dep_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Dep_hour'] = df['Dep_hour'].astype(int)
df['Dep_minute'] = df['Dep_minute'].astype(int)
```

```python
df.drop('Dep_Time', axis=1, inplace= True)
```

```python
df.head(2)
```

```python
df['Total_Stops'].unique()
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df['Total_Stops'] = df['Total_Stops'].map({
    'non-stop' : 0,
    '1 stop': 1,
    '2 stops': 2, 
    '3 stops' : 3, 
    '4 stops' : 4,
    np.nan : 1
})
```

```python
df['Total_Stops'].unique()
```

```python
df.head(2)
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
df['Duration_Hour'] = df['Duration'].str.split(' ').str[0].str.split('h').str[0]
df['Duration_Hour']
```

```python
df['Duration_Hour'].unique()
```

```python
df.drop(df[df['Duration_Hour'] == '5m'].index, inplace=True)
```

```python
df['Duration_Hour'].unique()
```

```python
df['Duration_Hour'] = df['Duration_Hour'].astype(int)
df['Duration_Hour']
```

```python
df['Duration_Hour_min'] = df['Duration_Hour']*60
df['Duration_Hour_min']
```

```python
df['Duration_min'] =df['Duration'].str.split(' ').str[1].str.split('m').str[0]
df['Duration_min']
```

```python
df['Duration_min'].unique()
```

```python
df['Duration_min'] = df['Duration_min'].fillna('0')
```

```python
df['Duration_min'].unique()
```

```python
df['Duration_min']= df['Duration_min'].astype(int)
```

```python
df['Total_Duration_Time'] = df['Duration_min'] + df['Duration_Hour_min']
```

```python
df.drop('Duration_Hour_min',axis=1,inplace=True)
df.drop('Duration_min',axis=1,inplace=True)
df.drop('Duration',axis=1,inplace=True)
df.drop('Duration_Hour',axis=1,inplace=True)
```
